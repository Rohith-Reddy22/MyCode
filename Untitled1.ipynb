```python
import json
import gc
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

SEED = 2026
rng = np.random.default_rng(SEED)

RUN_DIR = PREPARED_DIR
RUN_DIR.mkdir(exist_ok=True)

CLIENT_FILES = {
    "C1": DATA_DIR / "C1_data.csv",
    "C2": DATA_DIR / "C2_data.csv",
    "C3": DATA_DIR / "C3_data.csv",
}

TOTAL_ROWS = {"C1": 4_415_331, "C2": 17_251_169, "C3": 2_617_850}

CATEGORY_TO_SEMANTIC = {
    "C1": {0: "DDoS", 1: "Sinkhole", 2: "MiTM", 3: "Normal"},
    "C2": {0: "DDoS", 1: "DoS", 2: "Normal", 3: "SelectiveForwarding"},
    "C3": {0: "DDoS", 1: "DoS", 2: "Normal", 3: "Sybil"},
}

GLOBAL_CLASSES = ["Normal", "DDoS", "DoS", "MiTM", "SelectiveForwarding", "Sinkhole", "Sybil"]
CLASS_TO_ID = {name: index for index, name in enumerate(GLOBAL_CLASSES)}

MAX_ROWS_PER_CLIENT = 150_000
CHUNK_SIZE = 250_000
TEST_SIZE = 0.15
VALIDATION_SIZE = 0.15
Z_SCORE_CLIP = 10.0

FORBIDDEN_COLUMNS = {"label", "category", "type", "FPT", "LPT"}

header = pd.read_csv(CLIENT_FILES["C1"], nrows=0)
FEATURE_COLUMNS = [col for col in header.columns if col not in FORBIDDEN_COLUMNS]
USE_COLUMNS = FEATURE_COLUMNS + ["category"]

def stream_uniform_sample(client_name, csv_path):
    probability = min(1.0, MAX_ROWS_PER_CLIENT / TOTAL_ROWS[client_name])
    pieces = []
    for chunk in pd.read_csv(csv_path, usecols=USE_COLUMNS, chunksize=CHUNK_SIZE, low_memory=False):
        keep = rng.random(len(chunk)) < probability
        if keep.any():
            pieces.append(chunk.loc[keep])

    sampled = pd.concat(pieces, ignore_index=True)
    if len(sampled) > MAX_ROWS_PER_CLIENT:
        sampled = sampled.sample(n=MAX_ROWS_PER_CLIENT, random_state=SEED).reset_index(drop=True)

    semantic = sampled["category"].map(CATEGORY_TO_SEMANTIC[client_name])
    y = semantic.map(CLASS_TO_ID).to_numpy(dtype=np.int64)
    X = sampled[FEATURE_COLUMNS].apply(pd.to_numeric, errors="coerce").to_numpy(dtype=np.float32)

    del pieces, sampled, semantic
    gc.collect()
    return X, y

def local_split(X, y, client_name):
    indices = np.arange(len(y))
    train_idx, temp_idx = train_test_split(indices, test_size=TEST_SIZE + VALIDATION_SIZE, random_state=SEED, stratify=y)
    temp_y = y[temp_idx]
    val_frac = VALIDATION_SIZE / (TEST_SIZE + VALIDATION_SIZE)
    val_idx, test_idx = train_test_split(temp_idx, test_size=1 - val_frac, random_state=SEED, stratify=temp_y)
    return {"train": (X[train_idx], y[train_idx]), "val": (X[val_idx], y[val_idx]), "test": (X[test_idx], y[test_idx])}

raw_splits = {}
for client_name, csv_path in CLIENT_FILES.items():
    X_client, y_client = stream_uniform_sample(client_name, csv_path)
    raw_splits[client_name] = local_split(X_client, y_client, client_name)
    del X_client, y_client
    gc.collect()

feature_count = len(FEATURE_COLUMNS)
sum_values = np.zeros(feature_count, dtype=np.float64)
sum_squares = np.zeros(feature_count, dtype=np.float64)
valid_counts = np.zeros(feature_count, dtype=np.float64)

for client_name in CLIENT_FILES:
    X_train, _ = raw_splits[client_name]["train"]
    finite = np.isfinite(X_train)
    valid_counts += finite.sum(axis=0)
    sum_values += np.where(finite, X_train, 0.0).sum(axis=0, dtype=np.float64)
    sum_squares += np.where(finite, X_train * X_train, 0.0).sum(axis=0, dtype=np.float64)

mean = sum_values / valid_counts
variance = np.maximum((sum_squares / valid_counts) - (mean ** 2), 0.0)
std = np.sqrt(variance)

keep_mask = std > 1e-12
SELECTED_FEATURES = [f for f, keep in zip(FEATURE_COLUMNS, keep_mask) if keep]
mean, std = mean[keep_mask], std[keep_mask]

def normalize(X):
    X = X[:, keep_mask].astype(np.float32, copy=True)
    X = np.where(np.isfinite(X), X, mean[None, :])
    X = (X - mean[None, :]) / std[None, :]
    return np.clip(X, -Z_SCORE_CLIP, Z_SCORE_CLIP).astype(np.float32)

report_rows = []
for client_name in CLIENT_FILES:
    for split_name, (X, y) in raw_splits[client_name].items():
        X_prepared = normalize(X)
        np.savez_compressed(RUN_DIR / f"{client_name}_{split_name}.npz", X=X_prepared, y=y.astype(np.int64))
        for class_id, records in pd.Series(y).value_counts().items():
            report_rows.append({"client": client_name, "split": split_name, "class_id": int(class_id), "semantic_class": GLOBAL_CLASSES[int(class_id)], "records": int(records)})
        del X_prepared
    gc.collect()

pd.DataFrame(report_rows).sort_values(["client", "split", "class_id"]).to_csv(RUN_DIR / "split_class_distribution.csv", index=False)

metadata = {
    "seed": SEED, "global_classes": GLOBAL_CLASSES, "class_to_id": CLASS_TO_ID,
    "prepared_features": SELECTED_FEATURES
}
with open(RUN_DIR / "metadata.json", "w") as file: json.dump(metadata, file, indent=2)
print("Data preparation complete.")
```

```python
import os
import json
import time
import copy
import random
from pathlib import Path

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    matthews_corrcoef,
    confusion_matrix,
)

# ---------- Locked experiment configuration ----------
SEED = 2026
CLIENTS = ["C1", "C2", "C3"]

RUN_DIR = Path("/kaggle/working/flmu_final_run/prepared")
FS_DIR = Path("/kaggle/working/flmu_final_run/feature_selection")
RESULTS_DIR = Path("/kaggle/working/flmu_final_run/results")
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

BATCH_SIZE = 1024
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4
FOCAL_GAMMA = 2.0
MAX_CLASS_WEIGHT = 50.0

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

def seed_everything(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

seed_everything(SEED)

# ---------- Load class metadata and final feature indices ----------
with open(RUN_DIR / "metadata.json") as file:
    metadata = json.load(file)

GLOBAL_CLASSES = metadata["global_classes"]
NUM_CLASSES = len(GLOBAL_CLASSES)
NORMAL_CLASS_ID = GLOBAL_CLASSES.index("Normal")

selected_indices = np.load(FS_DIR / "selected_feature_indices.npy")

with open(FS_DIR / "selected_features.json") as file:
    selected_feature_info = json.load(file)

assert len(selected_indices) == 35
print("Classes:", GLOBAL_CLASSES)
print("Selected features:", len(selected_indices))

# ---------- Load the same selected schema for every method ----------
DATA = {}

for client in CLIENTS:
    DATA[client] = {}
    for split in ["train", "val", "test"]:
        loaded = np.load(RUN_DIR / f"{client}_{split}.npz")
        X = loaded["X"][:, selected_indices].astype(np.float32)
        y = loaded["y"].astype(np.int64)
        DATA[client][split] = (X, y)
        print(f"{client:>2} | {split:>5} | X={X.shape}, y={y.shape}")

# ---------- Lightweight IDS classifier: 35 -> 128 -> 64 -> 7 ----------
class LightweightMLP(nn.Module):
    def __init__(self, input_size, output_size):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(input_size, 128),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(0.10),
            nn.Linear(64, output_size),
        )

    def forward(self, x):
        return self.network(x)

def parameter_count(model):
    return sum(p.numel() for p in model.parameters())

temp_model = LightweightMLP(len(selected_indices), NUM_CLASSES)
print("Trainable parameters:", parameter_count(temp_model))

# ---------- Class-balanced focal loss ----------
def client_class_weights(y):
    counts = np.bincount(y, minlength=NUM_CLASSES).astype(np.float32)
    weights = np.zeros(NUM_CLASSES, dtype=np.float32)

    present = counts > 0
    weights[present] = counts[present].sum() / (
        present.sum() * counts[present]
    )

    weights = np.clip(weights, 0.0, MAX_CLASS_WEIGHT)
    return torch.tensor(weights, dtype=torch.float32, device=DEVICE)

def focal_loss(logits, targets, weights, gamma=FOCAL_GAMMA):
    ce = F.cross_entropy(logits, targets, weight=weights, reduction="none")
    probabilities = F.softmax(logits, dim=1)
    p_true = probabilities.gather(1, targets.unsqueeze(1)).squeeze(1)
    return ((1.0 - p_true).pow(gamma) * ce).mean()

def clone_state(model):
    return {
        key: value.detach().cpu().clone()
        for key, value in model.state_dict().items()
    }

def evaluate_model(model, X, y):
    model = model.to(DEVICE)
    model.eval()

    loader = DataLoader(
        TensorDataset(torch.from_numpy(X), torch.from_numpy(y)),
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
    )

    predictions = []

    with torch.no_grad():
        for X_batch, _ in loader:
            logits = model(X_batch.to(DEVICE, non_blocking=True))
            predictions.append(torch.argmax(logits, dim=1).cpu().numpy())

    model.to("cpu")
    return y, np.concatenate(predictions)

def calculate_metrics(y_true, y_pred):
    macro_p, macro_r, macro_f1, _ = precision_recall_fscore_support(
        y_true, y_pred, average="macro", zero_division=0
    )
    weighted_f1 = precision_recall_fscore_support(
        y_true, y_pred, average="weighted", zero_division=0
    )[2]

    normal_rows = y_true == NORMAL_CLASS_ID
    benign_to_attack_fpr = (
        float(np.mean(y_pred[normal_rows] != NORMAL_CLASS_ID))
        if normal_rows.any() else np.nan
    )

    return {
        "accuracy": float(accuracy_score(y_true, y_pred)),
        "macro_precision": float(macro_p),
        "macro_recall": float(macro_r),
        "macro_f1": float(macro_f1),
        "weighted_f1": float(weighted_f1),
        "mcc": float(matthews_corrcoef(y_true, y_pred)),
        "benign_to_attack_fpr": benign_to_attack_fpr,
    }

print("Setup complete.")
```

```python
# ==========================================================
# LOCKED MATCHED EXPERIMENTS
# Methods:
# Centralized | LocalOnly | FedAvg | FedProx |
# CB-FedProx | Proposed_CB_PFedProx
# ==========================================================

FEDERATED_ROUNDS = 30
CENTRAL_EPOCHS = 30
LOCAL_EPOCHS = 30
PERSONALIZATION_EPOCHS = 10
FEDPROX_MU = 0.01
PATIENCE = 6

EXPERIMENT_CONFIG = {
    "seed": SEED,
    "features": int(len(selected_indices)),
    "model": "MLP(35, 128, 64, 7)",
    "parameters": int(parameter_count(LightweightMLP(len(selected_indices), NUM_CLASSES))),
    "loss": "class-weighted focal loss",
    "focal_gamma": FOCAL_GAMMA,
    "max_class_weight": MAX_CLASS_WEIGHT,
    "learning_rate": LEARNING_RATE,
    "weight_decay": WEIGHT_DECAY,
    "batch_size": BATCH_SIZE,
    "federated_rounds": FEDERATED_ROUNDS,
    "fedprox_mu": FEDPROX_MU,
    "personalization_epochs": PERSONALIZATION_EPOCHS,
    "aggregation": {
        "FedAvg_and_FedProx": "training-sample-weighted",
        "CB_FedProx": "equal-client-weighted",
    },
}

with open(RESULTS_DIR / "experiment_config.json", "w") as file:
    json.dump(EXPERIMENT_CONFIG, file, indent=2)

def make_loader(X, y, shuffle=True):
    return DataLoader(
        TensorDataset(torch.from_numpy(X), torch.from_numpy(y)),
        batch_size=BATCH_SIZE,
        shuffle=shuffle,
        num_workers=0,
        pin_memory=torch.cuda.is_available(),
    )

def train_one_epoch(model, X, y, weights, reference_params=None, proximal_mu=0.0):
    model = model.to(DEVICE)
    model.train()

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY,
    )

    running_loss = 0.0
    total_examples = 0

    for X_batch, y_batch in make_loader(X, y, shuffle=True):
        X_batch = X_batch.to(DEVICE, non_blocking=True)
        y_batch = y_batch.to(DEVICE, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)
        loss = focal_loss(model(X_batch), y_batch, weights)

        if reference_params is not None and proximal_mu > 0:
            proximal_term = torch.zeros((), device=DEVICE)
            for name, parameter in model.named_parameters():
                proximal_term += torch.sum(
                    (parameter - reference_params[name]) ** 2
                )
            loss = loss + (proximal_mu / 2.0) * proximal_term

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
        optimizer.step()

        running_loss += loss.item() * len(y_batch)
        total_examples += len(y_batch)

    model.to("cpu")
    return running_loss / max(total_examples, 1)

def aggregate_models(local_models, client_weights):
    """Aggregate model parameters using supplied client weights."""
    global_model = LightweightMLP(len(selected_indices), NUM_CLASSES)
    aggregate_state = {}

    normalized_weights = np.asarray(client_weights, dtype=np.float64)
    normalized_weights = normalized_weights / normalized_weights.sum()

    for key in global_model.state_dict():
        aggregate_state[key] = sum(
            weight * model.state_dict()[key].detach().cpu().float()
            for model, weight in zip(local_models, normalized_weights)
        ).to(global_model.state_dict()[key].dtype)

    global_model.load_state_dict(aggregate_state)
    return global_model

def combined_validation(models_by_client):
    all_true, all_pred = [], []

    for client in CLIENTS:
        X_val, y_val = DATA[client]["val"]
        model = models_by_client[client]
        y_true, y_pred = evaluate_model(model, X_val, y_val)
        all_true.append(y_true)
        all_pred.append(y_pred)

    return calculate_metrics(np.concatenate(all_true), np.concatenate(all_pred))

def model_for_all_clients(model):
    return {client: model for client in CLIENTS}

def train_with_validation(
    model, X_train, y_train, X_val, y_val, epochs, weights,
    reference_params=None, proximal_mu=0.0, label=""
):
    best_f1 = -1.0
    best_state = clone_state(model)
    stale_epochs = 0
    history = []

    for epoch in range(1, epochs + 1):
        loss = train_one_epoch(
            model, X_train, y_train, weights,
            reference_params=reference_params,
            proximal_mu=proximal_mu,
        )

        y_true, y_pred = evaluate_model(model, X_val, y_val)
        metrics = calculate_metrics(y_true, y_pred)
        metrics["epoch"] = epoch
        metrics["loss"] = loss
        history.append(metrics)

        print(
            f"{label} | epoch {epoch:02d}/{epochs} | "
            f"loss={loss:.4f} | val_macro_f1={metrics['macro_f1']:.4f}"
        )

        if metrics["macro_f1"] > best_f1:
            best_f1 = metrics["macro_f1"]
            best_state = clone_state(model)
            stale_epochs = 0
        else:
            stale_epochs += 1
            if stale_epochs >= PATIENCE:
                print(f"{label} | early stop at epoch {epoch}")
                break

    model.load_state_dict(best_state)
    return model, history

def run_federated(method_name, aggregation="sample", fedprox_mu=0.0):
    """
    aggregation='sample' uses client training-set size.
    aggregation='equal' gives every client equal aggregation weight.
    """
    seed_everything(SEED)

    global_model = LightweightMLP(len(selected_indices), NUM_CLASSES)
    best_f1 = -1.0
    best_state = clone_state(global_model)
    stale_rounds = 0
    history = []
    start_time = time.time()

    if aggregation == "sample":
        aggregation_weights = [
            len(DATA[client]["train"][1]) for client in CLIENTS
        ]
    elif aggregation == "equal":
        aggregation_weights = [1.0] * len(CLIENTS)
    else:
        raise ValueError("aggregation must be 'sample' or 'equal'")

    for round_number in range(1, FEDERATED_ROUNDS + 1):
        local_models = []

        for client in CLIENTS:
            X_train, y_train = DATA[client]["train"]

            local_model = copy.deepcopy(global_model)
            reference_params = None

            if fedprox_mu > 0:
                reference_params = {
                    name: parameter.detach().clone().to(DEVICE)
                    for name, parameter in local_model.named_parameters()
                }

            weights = client_class_weights(y_train)

            train_one_epoch(
                local_model, X_train, y_train, weights,
                reference_params=reference_params,
                proximal_mu=fedprox_mu,
            )
            local_models.append(local_model)

        global_model = aggregate_models(local_models, aggregation_weights)

        validation = combined_validation(model_for_all_clients(global_model))
        validation["round"] = round_number
        history.append(validation)

        print(
            f"{method_name} | round {round_number:02d}/{FEDERATED_ROUNDS} | "
            f"val_macro_f1={validation['macro_f1']:.4f} | "
            f"val_accuracy={validation['accuracy']:.4f}"
        )

        if validation["macro_f1"] > best_f1:
            best_f1 = validation["macro_f1"]
            best_state = clone_state(global_model)
            stale_rounds = 0
        else:
            stale_rounds += 1
            if stale_rounds >= PATIENCE:
                print(f"{method_name} | early stop at round {round_number}")
                break

    global_model.load_state_dict(best_state)
    elapsed = time.time() - start_time

    model_size_mb = (
        parameter_count(global_model) * 4 / (1024 ** 2)
    )

    # Each round: server sends one model to each client and receives one update.
    rounds_completed = len(history)
    communication_mb = 2 * len(CLIENTS) * rounds_completed * model_size_mb

    return global_model, history, elapsed, rounds_completed, communication_mb

def personalize_model(global_model, client):
    """Fine-tune only with that client's training data; select on local validation."""
    X_train, y_train = DATA[client]["train"]
    X_val, y_val = DATA[client]["val"]

    local_model = copy.deepcopy(global_model)
    weights = client_class_weights(y_train)

    model, history = train_with_validation(
        local_model,
        X_train, y_train,
        X_val, y_val,
        epochs=PERSONALIZATION_EPOCHS,
        weights=weights,
        label=f"Personalization-{client}",
    )
    return model, history

def collect_test_results(method_name, models_by_client,
                         training_seconds, communication_rounds,
                         communication_mb):
    summary_rows = []
    class_rows = []
    confusion_matrices = {}

    all_true, all_pred = [], []

    for client in CLIENTS:
        X_test, y_test = DATA[client]["test"]
        y_true, y_pred = evaluate_model(models_by_client[client], X_test, y_test)

        metrics = calculate_metrics(y_true, y_pred)
        summary_rows.append({
            "method": method_name,
            "evaluation_client": client,
            **metrics,
            "training_seconds": training_seconds,
            "communication_rounds": communication_rounds,
            "communication_mb": communication_mb,
            "parameters": parameter_count(models_by_client[client]),
        })

        precision, recall, f1, support = precision_recall_fscore_support(
            y_true,
            y_pred,
            labels=list(range(NUM_CLASSES)),
            zero_division=0,
        )

        for class_id, class_name in enumerate(GLOBAL_CLASSES):
            class_rows.append({
                "method": method_name,
                "evaluation_client": client,
                "class_id": class_id,
                "class_name": class_name,
                "precision": float(precision[class_id]),
                "recall": float(recall[class_id]),
                "f1_score": float(f1[class_id]),
                "support": int(support[class_id]),
            })

        confusion_matrices[f"{method_name}_{client}"] = confusion_matrix(
            y_true, y_pred, labels=list(range(NUM_CLASSES))
        )

        all_true.append(y_true)
        all_pred.append(y_pred)

    all_true = np.concatenate(all_true)
    all_pred = np.concatenate(all_pred)
    all_metrics = calculate_metrics(all_true, all_pred)

    summary_rows.append({
        "method": method_name,
        "evaluation_client": "AllClients",
        **all_metrics,
        "training_seconds": training_seconds,
        "communication_rounds": communication_rounds,
        "communication_mb": communication_mb,
        "parameters": parameter_count(next(iter(models_by_client.values()))),
    })

    return summary_rows, class_rows, confusion_matrices

# ----------------------------------------------------------
# 1. Centralized baseline
# ----------------------------------------------------------
seed_everything(SEED)
central_start = time.time()

X_central_train = np.concatenate(
    [DATA[client]["train"][0] for client in CLIENTS]
)
y_central_train = np.concatenate(
    [DATA[client]["train"][1] for client in CLIENTS]
)
X_central_val = np.concatenate(
    [DATA[client]["val"][0] for client in CLIENTS]
)
y_central_val = np.concatenate(
    [DATA[client]["val"][1] for client in CLIENTS]
)

central_model = LightweightMLP(len(selected_indices), NUM_CLASSES)
central_model, central_history = train_with_validation(
    central_model,
    X_central_train, y_central_train,
    X_central_val, y_central_val,
    epochs=CENTRAL_EPOCHS,
    weights=client_class_weights(y_central_train),
    label="Centralized",
)
central_seconds = time.time() - central_start

# ----------------------------------------------------------
# 2. Local-only baseline
# ----------------------------------------------------------
seed_everything(SEED)
local_start = time.time()
local_models = {}
local_history = {}

for client in CLIENTS:
    X_train, y_train = DATA[client]["train"]
    X_val, y_val = DATA[client]["val"]

    local_model = LightweightMLP(len(selected_indices), NUM_CLASSES)
    local_models[client], local_history[client] = train_with_validation(
        local_model,
        X_train, y_train,
        X_val, y_val,
        epochs=LOCAL_EPOCHS,
        weights=client_class_weights(y_train),
        label=f"LocalOnly-{client}",
    )

local_seconds = time.time() - local_start

# ----------------------------------------------------------
# 3. Sample-weighted FedAvg
# ----------------------------------------------------------
fedavg_model, fedavg_history, fedavg_seconds, fedavg_rounds, fedavg_comm = (
    run_federated(
        method_name="FedAvg",
        aggregation="sample",
        fedprox_mu=0.0,
    )
)

# ----------------------------------------------------------
# 4. Sample-weighted FedProx
# ----------------------------------------------------------
fedprox_model, fedprox_history, fedprox_seconds, fedprox_rounds, fedprox_comm = (
    run_federated(
        method_name="FedProx",
        aggregation="sample",
        fedprox_mu=FEDPROX_MU,
    )
)

# ----------------------------------------------------------
# 5. Equal-client FedProx ablation
# ----------------------------------------------------------
cb_fedprox_model, cb_history, cb_seconds, cb_rounds, cb_comm = run_federated(
    method_name="CB-FedProx",
    aggregation="equal",
    fedprox_mu=FEDPROX_MU,
)

# ----------------------------------------------------------
# 6. Proposed method: equal-client FedProx + personalization
# ----------------------------------------------------------
personal_start = time.time()
personalized_models = {}
personal_history = {}

for client in CLIENTS:
    personalized_models[client], personal_history[client] = personalize_model(
        cb_fedprox_model, client
    )

proposed_seconds = cb_seconds + (time.time() - personal_start)

# ----------------------------------------------------------
# Save checkpoints
# ----------------------------------------------------------
torch.save(central_model.state_dict(), RESULTS_DIR / "centralized_model.pt")
torch.save(fedavg_model.state_dict(), RESULTS_DIR / "fedavg_model.pt")
torch.save(fedprox_model.state_dict(), RESULTS_DIR / "fedprox_model.pt")
torch.save(cb_fedprox_model.state_dict(), RESULTS_DIR / "cb_fedprox_model.pt")

for client, model in local_models.items():
    torch.save(model.state_dict(), RESULTS_DIR / f"localonly_{client}.pt")

for client, model in personalized_models.items():
    torch.save(model.state_dict(), RESULTS_DIR / f"proposed_personalized_{client}.pt")

# ----------------------------------------------------------
# Collect and save final test results
# ----------------------------------------------------------
all_summary_rows = []
all_class_rows = []
all_confusions = {}

experiment_outputs = [
    (
        "Centralized",
        model_for_all_clients(central_model),
        central_seconds, 0, 0.0,
    ),
    (
        "LocalOnly",
        local_models,
        local_seconds, 0, 0.0,
    ),
    (
        "FedAvg",
        model_for_all_clients(fedavg_model),
        fedavg_seconds, fedavg_rounds, fedavg_comm,
    ),
    (
        "FedProx",
        model_for_all_clients(fedprox_model),
        fedprox_seconds, fedprox_rounds, fedprox_comm,
    ),
    (
        "CB-FedProx",
        model_for_all_clients(cb_fedprox_model),
        cb_seconds, cb_rounds, cb_comm,
    ),
    (
        "Proposed_CB_PFedProx",
        personalized_models,
        proposed_seconds, cb_rounds, cb_comm,
    ),
]

for method_name, models, seconds, rounds, communication in experiment_outputs:
    summary, class_metrics, confusions = collect_test_results(
        method_name,
        models,
        seconds,
        rounds,
        communication,
    )
    all_summary_rows.extend(summary)
    all_class_rows.extend(class_metrics)
    all_confusions.update(confusions)

summary_df = pd.DataFrame(all_summary_rows)
class_df = pd.DataFrame(all_class_rows)

summary_df.to_csv(RESULTS_DIR / "main_results.csv", index=False)
class_df.to_csv(RESULTS_DIR / "per_class_results.csv", index=False)
np.savez_compressed(
    RESULTS_DIR / "confusion_matrices.npz",
    **all_confusions,
)

pd.DataFrame(central_history).to_csv(
    RESULTS_DIR / "history_centralized.csv", index=False
)
pd.DataFrame(fedavg_history).to_csv(
    RESULTS_DIR / "history_fedavg.csv", index=False
)
pd.DataFrame(fedprox_history).to_csv(
    RESULTS_DIR / "history_fedprox.csv", index=False
)
pd.DataFrame(cb_history).to_csv(
    RESULTS_DIR / "history_cb_fedprox.csv", index=False
)

for client, history in local_history.items():
    pd.DataFrame(history).to_csv(
        RESULTS_DIR / f"history_localonly_{client}.csv", index=False
    )

for client, history in personal_history.items():
    pd.DataFrame(history).to_csv(
        RESULTS_DIR / f"history_personalization_{client}.csv", index=False
    )

display_columns = [
    "method",
    "accuracy",
    "macro_precision",
    "macro_recall",
    "macro_f1",
    "weighted_f1",
    "mcc",
    "benign_to_attack_fpr",
    "communication_rounds",
    "communication_mb",
]

final_table = (
    summary_df[summary_df["evaluation_client"] == "AllClients"]
    .sort_values("macro_f1", ascending=False)
    [display_columns]
    .reset_index(drop=True)
)

print("\n========== FINAL ALL-CLIENT TEST RESULTS ==========")
display(final_table)

print("\nSaved outputs to:", RESULTS_DIR)
print("Do not tune the configuration further using these test results.")
```
